## 8. Menentukan Jumlah Cluster Optimal

Jumlah cluster pada K-Means dianalisis menggunakan dua pendekatan:

1. Elbow Method
2. Silhouette Score

Elbow Method menggunakan nilai inertia untuk melihat perubahan total
within-cluster sum of squares terhadap jumlah cluster.

Silhouette Score digunakan untuk melihat kualitas pemisahan data antar
cluster.

```python
k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    inertias.append(model.inertia_)

    silhouette_scores.append(
        silhouette_score(X_scaled, labels)
    )

print("Perhitungan K-Means untuk berbagai nilai K selesai.")
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    inertias,
    marker="o"
)

plt.title("Elbow Method")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Inertia")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score untuk Berbagai K")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Silhouette Score")

plt.tight_layout()
plt.show()
```

```python
k_results = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertias,
    "Silhouette_Score": silhouette_scores
})

k_results
```

```python
best_k_silhouette = int(
    k_results.loc[
        k_results["Silhouette_Score"].idxmax(),
        "K"
    ]
)

print(
    "K dengan Silhouette Score tertinggi:",
    best_k_silhouette
)
```

## 9. Pemilihan Jumlah Cluster

Berdasarkan hasil pengujian K = 2 sampai K = 10, nilai Silhouette Score
tertinggi diperoleh pada K = 2 dengan nilai 0.324881.

K = 3 menghasilkan Silhouette Score sebesar 0.315327, yang masih relatif
dekat dengan K = 2, tetapi tetap lebih rendah.

Pada K >= 4, Silhouette Score menurun secara cukup signifikan dan berada
pada kisaran sekitar 0.19 hingga 0.21.

Berdasarkan pertimbangan Silhouette Score dan pemeriksaan Elbow Method,
eksperimen utama K-Means menggunakan K = 2.

Dengan demikian, dataset dikelompokkan menjadi dua cluster:
Cluster 0 dan Cluster 1.

```python
# Setelah melihat Elbow Method dan Silhouette Score,
# tetapkan nilai K yang dipilih di sini.

OPTIMAL_K = best_k_silhouette

print("K yang digunakan:", OPTIMAL_K)
```

## 10. K-Means Clustering

K-Means digunakan untuk membagi data menjadi sejumlah cluster berdasarkan
kedekatan data terhadap centroid.

Model dilatih menggunakan data yang telah melalui StandardScaler.

```python
kmeans = KMeans(
    n_clusters=OPTIMAL_K,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)

print("K-Means berhasil dilatih.")
print("Jumlah cluster:", OPTIMAL_K)
```

## 11. Agglomerative Clustering

Agglomerative Clustering digunakan sebagai algoritma hierarchical
clustering untuk membandingkan hasil dengan K-Means.

Metode `ward` digunakan sesuai dengan ketentuan tugas.

```python
agglomerative = AgglomerativeClustering(
    n_clusters=OPTIMAL_K,
    linkage="ward"
)

agglomerative_labels = agglomerative.fit_predict(X_scaled)

print("Agglomerative Clustering berhasil dilatih.")
```

## 12. DBSCAN

DBSCAN merupakan algoritma clustering berbasis kepadatan.

Parameter utama:

- `eps`: radius pencarian tetangga
- `min_samples`: jumlah minimum titik untuk membentuk area padat

Data yang diberi label `-1` dianggap sebagai noise.

```python
dbscan_params = [
    (0.5, 5),
    (0.7, 5),
    (0.9, 5),
    (1.1, 5),
    (1.3, 5),
    (0.7, 10),
    (0.9, 10),
    (1.1, 10)
]

dbscan_results = []

for eps, min_samples in dbscan_params:

    model = DBSCAN(
        eps=eps,
        min_samples=min_samples
    )

    labels = model.fit_predict(X_scaled)

    n_clusters = len(
        set(labels) - {-1}
    )

    noise_pct = (
        np.sum(labels == -1)
        / len(labels)
        * 100
    )

    non_noise = labels != -1

    if (
        n_clusters >= 2
        and np.sum(non_noise) > 1
        and len(set(labels[non_noise])) >= 2
    ):

        sil = silhouette_score(
            X_scaled[non_noise],
            labels[non_noise]
        )

        db = davies_bouldin_score(
            X_scaled[non_noise],
            labels[non_noise]
        )

        ch = calinski_harabasz_score(
            X_scaled[non_noise],
            labels[non_noise]
        )

    else:
        sil = np.nan
        db = np.nan
        ch = np.nan

    dbscan_results.append({
        "eps": eps,
        "min_samples": min_samples,
        "Jumlah Cluster": n_clusters,
        "Silhouette Score": sil,
        "Davies-Bouldin": db,
        "Calinski-Harabasz": ch,
        "Noise (%)": noise_pct
    })

dbscan_results = pd.DataFrame(
    dbscan_results
)

dbscan_results
```

```python
valid_dbscan = dbscan_results.dropna(
    subset=["Silhouette Score"]
)

best_dbscan = valid_dbscan.loc[
    valid_dbscan["Silhouette Score"].idxmax()
]

BEST_EPS = float(best_dbscan["eps"])
BEST_MIN_SAMPLES = int(
    best_dbscan["min_samples"]
)

print("Parameter DBSCAN:")
print("eps:", BEST_EPS)
print("min_samples:", BEST_MIN_SAMPLES)
```

```python
dbscan = DBSCAN(
    eps=BEST_EPS,
    min_samples=BEST_MIN_SAMPLES
)

dbscan_labels = dbscan.fit_predict(X_scaled)

n_dbscan_clusters = len(
    set(dbscan_labels) - {-1}
)

dbscan_noise_pct = (
    np.mean(dbscan_labels == -1)
    * 100
)

print("DBSCAN berhasil dilatih.")
print("Jumlah cluster:", n_dbscan_clusters)
print("Noise:", round(dbscan_noise_pct, 2), "%")
```

## 13. Evaluasi Clustering

Ketiga algoritma dibandingkan menggunakan metrik internal clustering:

- Silhouette Score
- Davies-Bouldin Index
- Calinski-Harabasz Index
- Jumlah cluster
- Noise percentage

Untuk DBSCAN, titik dengan label `-1` diperlakukan sebagai noise.

```python
def evaluate_clustering(
    X_data,
    labels,
    ignore_noise=False
):
    labels = np.asarray(labels)

    if ignore_noise:
        mask = labels != -1
        X_eval = X_data[mask]
        labels_eval = labels[mask]
    else:
        X_eval = X_data
        labels_eval = labels

    unique_labels = set(labels_eval)

    if len(unique_labels) < 2:
        return {
            "Jumlah Cluster": len(unique_labels),
            "Silhouette Score": np.nan,
            "Davies-Bouldin": np.nan,
            "Calinski-Harabasz": np.nan,
            "Noise (%)": np.mean(labels == -1) * 100
        }

    return {
        "Jumlah Cluster": len(unique_labels),
        "Silhouette Score": silhouette_score(
            X_eval,
            labels_eval
        ),
        "Davies-Bouldin": davies_bouldin_score(
            X_eval,
            labels_eval
        ),
        "Calinski-Harabasz": calinski_harabasz_score(
            X_eval,
            labels_eval
        ),
        "Noise (%)": (
            np.mean(labels == -1) * 100
        )
    }
```

```python
kmeans_eval = evaluate_clustering(
    X_scaled,
    kmeans_labels
)

agglomerative_eval = evaluate_clustering(
    X_scaled,
    agglomerative_labels
)

dbscan_eval = evaluate_clustering(
    X_scaled,
    dbscan_labels,
    ignore_noise=True
)

comparison = pd.DataFrame(
    [
        kmeans_eval,
        agglomerative_eval,
        dbscan_eval
    ],
    index=[
        "K-Means",
        "Agglomerative",
        "DBSCAN"
    ]
)

comparison
```

## 15. Principal Component Analysis (PCA)

PCA digunakan untuk mereduksi dimensi data.

Analisis pertama dilakukan menggunakan seluruh principal component untuk
mengetahui berapa banyak component yang diperlukan untuk menangkap
minimal 80% varians data.

PCA kemudian digunakan untuk membuat visualisasi dua dimensi hasil clustering.

```python
pca_full = PCA()

X_pca_full = pca_full.fit_transform(
    X_scaled
)

explained_variance = (
    pca_full.explained_variance_ratio_
)

cumulative_variance = np.cumsum(
    explained_variance
)

n_components_80 = (
    np.argmax(
        cumulative_variance >= 0.80
    )
    + 1
)

print(
    "Jumlah component untuk >= 80% varians:",
    n_components_80
)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    range(
        1,
        len(cumulative_variance) + 1
    ),
    cumulative_variance,
    marker="o"
)

plt.axhline(
    y=0.80,
    linestyle="--"
)

plt.axvline(
    x=n_components_80,
    linestyle="--"
)

plt.title(
    "Cumulative Explained Variance PCA"
)

plt.xlabel(
    "Jumlah Principal Component"
)

plt.ylabel(
    "Cumulative Explained Variance"
)

plt.tight_layout()
plt.show()
```

## 16. PCA 2D

Dua principal component pertama digunakan untuk membuat visualisasi
hasil clustering dalam ruang dua dimensi.

Persentase explained variance dari PC1 dan PC2 ditampilkan pada label
sumbu.

```python
pca_2d = PCA(
    n_components=2,
    random_state=42
)

X_pca_2d = pca_2d.fit_transform(
    X_scaled
)

pc1_variance = (
    pca_2d.explained_variance_ratio_[0]
    * 100
)

pc2_variance = (
    pca_2d.explained_variance_ratio_[1]
    * 100
)

print(
    f"PC1 menjelaskan {pc1_variance:.2f}% varians"
)

print(
    f"PC2 menjelaskan {pc2_variance:.2f}% varians"
)
```
